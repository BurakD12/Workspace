```python
'''
    Operators

    1. Arithmetic Operators
    2. Assignment Operators
    3. Comparison Operators
    4. Logical Operators
    5. Bitwise Operators
    6. Special Operators
'''

'''
    1. Python Arithmetic Operators

        Operator	    Operation	            Example
        +	            Addition	            5 + 2 = 7
        -	            Subtraction	            4 - 2 = 2
        *	            Multiplication	        2 * 3 = 6
        /	            Division	            4 / 2 = 2
        //	            Floor Division	        10 // 3 = 3
        %	            Modulo	                5 % 2 = 1
        **	            Power	                4 ** 2 = 16 
'''
```

```python
a = 7
b = 2

print ('Sum: ', a + b)  #9
print ('Subtraction: ', a - b) #5   
print ('Multiplication: ', a * b) #14 
print ('Division: ', a / b) #3.5
print ('Floor Division: ', a // b) #3
print ('Modulo: ', a % b)  #1
print ('Power: ', a ** b)   #49
```

```python
'''
    2. Python Assignment Operators

    Operator	    Name	                        Example
        =	        Assignment Operator	            a = 7
        +=	        Addition Assignment	            a += 1               a = a + 1
        -=	        Subtraction Assignment	        a -= 3               a = a - 3
        *=	        Multiplication Assignment	    a *= 4               a = a * 4
        /=	        Division Assignment	            a /= 3               a = a / 3
        %=	        Remainder Assignment	        a %= 10              a = a % 10
        **=	        Exponent Assignment	            a **= 10             a = a ** 10
'''
```

```python
'''
    3. Python Comparison Operators

    Operator	        Meaning	                        Example
    ==	                Is Equal To	                    3 == 5 gives us False
    !=	                Not Equal To	                3 != 5 gives us True
    >	                Greater Than	                3 > 5 gives us False
    <	                Less Than	                    3 < 5 gives us True
    >=	                Greater Than or Equal To	    3 >= 5 give us False
    <=	                Less Than or Equal To	        3 <= 5 gives us True
'''

a = 5
b = 2

print('a == b =', a == b) #F
print('a != b =', a != b) #T
print('a > b =', a > b) #T
print('a < b =', a < b) #F
print('a >= b =', a >= b) #T
print('a <= b =', a <= b) #F
```

```python
'''
    4. Python Logical Operators

        Operator	        Example	              Meaning
        and	                a and b	              Logical AND: True only if both the operands are True
        or	                a or b	              Logical OR: True if at least one of the operands is True
        not	                not a	              Logical NOT: True if the operand is False and vice-versa.
'''

print(True and True)    
print(True and False)    
print(True or False)    
print(not True)        
```

```python
'''
    6. Python Special operators

        1. Identity Operators

            Operator	    Meaning	                                                                                    Example
            is	            True if the operands are identical (refer to the same object)	                            x is True
            is not	        True if the operands are not identical (do not refer to the same object)	                x is not True
'''

x1 = 5
y1 = 5
x2 = 'Hello'
y2 = 'Hello'
x3 = [1,2,3] # LIST@5454465
y3 = [1,2,3] # LIST@5434234

print(x1 is not y1) # F
print(x2 is y2) # T
print(x3 is y3) # F
```

```python
'''
    2. Membership Operators

        Operator	        Meaning                                             	        Example
        in	                True if value/variable is found in the sequence	                5 in x
        not in	            True if value/variable is not found in the sequence	            5 not in x
'''

message = 'Hello world'
dict1 = {1:'a', 2:'b'}

print('H' in message)  #T
print('hello' not in message) #T 
print(1 in dict1)  #T
print('a' in dict1) #F
```
