```python
# Let user enter a number and then find out sum of all numbers between 0-N

n = int(input('Enter your number'))
sum = 0

for i in range(0,n + 1):
    sum += i
    
sum
```

```python
# Let user enter a number and then find out its factorial.

n = int(input("Enter your number: "))
product = 1

for i in range(n, 0, -1):
    product *= i
    
product
```

```python
# Let user enter 2 numbers and then find the GCD of those 2 number

x = int(input("Enter number 1:"))
y = int(input("Enter number 2:"))

minNum = min(x,y)

for i in range(minNum,0,-1):
    if x % i == 0 and y % i == 0:
        print('Answer:',i)
        break
```
